```python
# Eliminar filas con NaN 
ls1 = ls1.dropna(subset=['host_is_superhost'])

# verificar cuántas filas quedaron
print("Filas después de eliminar NaN:", len(ls1))
```

```python
# Convertir la columna 'price' a string
ls1['price'] = ls1['price'].astype(str)

# Eliminar el símbolo '$'
ls1['price'] = ls1['price'].str.replace('$', '', regex=False)

# Quitar las comas
ls1['price'] = ls1['price'].str.replace(',', '', regex=False)

# Quitar decimales '.00'
ls1['price'] = ls1['price'].str.replace('.00', '', regex=False)

# Convertir a entero
ls1['price'] = ls1['price'].astype(int)

# Ver resultados
print(ls1['price'])
```

```python
ls1 = ls1.rename(columns={
    'host_is_superhost': 'Superhost',
    'neighbourhood_cleansed': 'Neighbourhood',
    'property_type': 'Property type',
    'room_type': 'Room type',
    'price': 'Price',
    'calendar_updated': 'Calendar updated',
    'has_availability': 'Availability',
    'number_of_reviews_ltm': 'Number of reviews from the last 12 months'
})
```

```python
# Boxplot para 'Price'
plt.figure(figsize=(10, 5))
sns.boxplot(x=ls1['Price'])
plt.title('Boxplot of Price')
plt.show()

# Boxplot para 'Number of reviews from the last 12 months'
plt.figure(figsize=(10, 5))
sns.boxplot(x=ls1['Number of reviews from the last 12 months'])
plt.title('Boxplot of Number of reviews from the last 12 months')
plt.show()
```

```python
# Calcular z-scores para 'Price'
z_scores_price = np.abs((ls1['Price'] - ls1['Price'].mean()) / ls1['Price'].std())

# Calcular z-scores para 'Number of reviews from the last 12 months'
z_scores_reviews = np.abs((ls1['Number of reviews from the last 12 months'] - 
                           ls1['Number of reviews from the last 12 months'].mean()) / 
                           ls1['Number of reviews from the last 12 months'].std())

# Identificar outliers con z-scores > 3
outliers_price = ls1[z_scores_price > 3]
outliers_reviews = ls1[z_scores_reviews > 3]

print("Outliers in Price:")
print(outliers_price)

print("\nOutliers in Number of reviews from the last 12 months:")
print(outliers_reviews)
```

```python
# Eliminar outliers en 'Price'
ls2 = ls1[z_scores_price <= 3]

# Eliminar outliers en 'Number of reviews from the last 12 months'
ls2 = ls1[z_scores_reviews <= 3]
ls2
```

```python
# --- 1. Define el umbral máximo de precio que consideras aceptable para registros INDIVIDUALES ---
# Este es el 'corte final' para cada fila en ls3.
MAX_PRICE_FINAL_THRESHOLD = 4000 # <--- Tu umbral de 4000

# --- 2. Filtra ls3 para eliminar las filas donde el 'Price' individual supera el umbral ---
# El resultado lo guardamos en ls3, que contendrá TODAS las columnas de ls2 y solo las filas filtradas.
ls3 = ls2[ls2['Price'] <= MAX_PRICE_FINAL_THRESHOLD].copy()

# --- Verificar los resultados ---
print(f"Número de filas en ls2 original (antes de este último filtro): {len(ls2)}")
print(f"Número de filas en ls3 (después del filtro final de precio individual): {len(ls3)}")
print(f"Filas eliminadas en este último paso: {len(ls2) - len(ls3)}\n")

print(f"Columnas en ls3: {list(ls3.columns)}") # Muestra todas las columnas de ls3

print("\n--- Verificación de los tipos de propiedad problemáticos en ls3 ---")
# La lista de tipos problemáticos (ahora veremos si tienen registros individuales por debajo del umbral)
problematic_types = ['Earth house', 'Hotel', 'Other', 'Townhouse', 'Villa', 'Boutique hotel', 'Aparthotel']

for prop_type in problematic_types:
    # Filtramos en ls3 para ver qué registros individuales de Property type están presentes y cumplen el umbral.
    filtered_rows = ls3[ls3['Property type'] == prop_type]
    if not filtered_rows.empty:
        print(f"\n{prop_type} (registros individuales presentes en ls3 con Price <= {MAX_PRICE_FINAL_THRESHOLD}):")
        # Mostrar algunas filas y sus precios para no inundar la consola
        print(filtered_rows[['Room type', 'Price']].head(5).to_string(index=False))
        if len(filtered_rows) > 5:
            print(f"... y {len(filtered_rows) - 5} más")
    else:
        print(f"\n{prop_type} (no hay registros individuales en ls3 con este Property type por debajo de {MAX_PRICE_FINAL_THRESHOLD})")

print(f"\nNuevo promedio máximo de Price en ls3 (de registros individuales): {ls3['Price'].max():.2f}")
print(f"Nuevo promedio general de Price en ls3 (de registros individuales): {ls3['Price'].mean():.2f}")


# --- IMPORTANTE: Si después quieres la tabla de promedios agrupados *a partir de este ls3*, la generarías así: ---
ls3_promedios_agrupados = ls3.groupby(['Property type', 'Room type'])['Price'].mean().reset_index(name='Average price')
print("\n--- Promedios agrupados calculados a partir del nuevo ls3 (Top 10) ---")
print(ls3_promedios_agrupados.sort_values(by='Average price', ascending=False).head(10).to_string(index=False))

# Y también puedes verificar qué promedios de Property Type quedan ahora:
ls3_promedios_property_type = ls3.groupby('Property type')['Price'].mean().reset_index(name='Average price')
print("\n--- Promedios de Property Type a partir del nuevo ls3 (Top 10) ---")
print(ls3_promedios_property_type.sort_values(by='Average price', ascending=False).head(10).to_string(index=False))
```

```python
# --- 1. IDENTIFICAR LOS REGISTROS ESPECÍFICOS A ELIMINAR ---

# Caso 1: Velez Sarsfield Shared room 3319
velez_sarsfield_problem_rows = ls3[
    (ls3['Neighbourhood'] == 'Velez Sarsfield') &
    (ls3['Room type'] == 'Shared room') &
    (ls3['Price'] == 3319)
]

# Caso 2: Villa Soldati Private room 3983
villa_soldati_problem_rows = ls3[
    (ls3['Neighbourhood'] == 'Villa Soldati') &
    (ls3['Room type'] == 'Private room') &
    (ls3['Price'] == 3983)
]

# Caso 3: Campsite Entire home/apt 3983
# Esto asume que 'Campsite' es un 'Property type'. Si fuera 'Neighbourhood', ajusta la condición.
# Si 3983 es un promedio de un grupo, estamos buscando las filas INDIVIDUALES que lo componen.
# Si este 3983 es el promedio de un grupo "Campsite", queremos eliminar los valores individuales
# de Price que componen ese promedio. Sin más contexto, asumiré que estamos buscando
# un Property type 'Campsite' y un Room type 'Entire home/apt' con Price = 3983.
campsite_problem_rows = ls3[
    (ls3['Property type'] == 'Campsite') &
    (ls3['Room type'] == 'Entire home/apt') &
    (ls3['Price'] == 3983)
    ]


# Unir los índices de todas las filas problemáticas
indices_to_remove = pd.Index([])
if not velez_sarsfield_problem_rows.empty:
    indices_to_remove = indices_to_remove.union(velez_sarsfield_problem_rows.index)
if not villa_soldati_problem_rows.empty:
    indices_to_remove = indices_to_remove.union(villa_soldati_problem_rows.index)
if not campsite_problem_rows.empty:
    indices_to_remove = indices_to_remove.union(campsite_problem_rows.index)

print(f"Número de registros identificados para eliminar: {len(indices_to_remove)}")
if not indices_to_remove.empty:
    print("Índices a eliminar:", indices_to_remove.tolist())
else:
    print("No se encontraron los registros problemáticos específicos.")

# --- 2. ELIMINAR ESOS REGISTROS DE LS5 ---
ls3_final = ls3.drop(indices_to_remove)

# --- 3. VERIFICAR LOS RESULTADOS FINALMENTE ---
print(f"\n--- VERIFICACIÓN DESPUÉS DE LA ELIMINACIÓN EXPLÍCITA ---")
print(f"Número de filas en ls3 antes de esta eliminación: {len(ls3)}")
print(f"Número de filas en ls3_final: {len(ls3_final)}")
print(f"Filas eliminadas en este paso: {len(ls3) - len(ls3_final)}\n")


# Verificar si los promedios de esos grupos han cambiado en ls3_final
print("Promedio de 'Velez Sarsfield', 'Shared room' en ls3_final:")
velez_avg_new = ls3_final[
    (ls3_final['Neighbourhood'] == 'Velez Sarsfield') &
    (ls3_final['Room type'] == 'Shared room')
]['Price'].mean()
print(f"  {velez_avg_new:.2f}" if not pd.isna(velez_avg_new) else "  (Grupo eliminado o ya no tiene registros)")

print("\nPromedio de 'Villa Soldati', 'Private room' en ls3_final:")
villa_avg_new = ls3_final[
    (ls3_final['Neighbourhood'] == 'Villa Soldati') &
     (ls3_final['Room type'] == 'Private room')
]['Price'].mean()
print(f"  {villa_avg_new:.2f}" if not pd.isna(villa_avg_new) else "  (Grupo eliminado o ya no tiene registros)")

print("\nPromedio de 'Campsite', 'Entire home/apt' en ls3_final:")
campsite_avg_new = ls3_final[
    (ls3_final['Property type'] == 'Campsite') &
    (ls3_final['Room type'] == 'Entire home/apt')
]['Price'].mean()
print(f"  {campsite_avg_new:.2f}" if not pd.isna(campsite_avg_new) else "  (Grupo eliminado o ya no tiene registros)")


# --- Opcional: Volver a calcular los promedios agrupados a partir de ls3_final ---
ls3_final_promedios_agrupados = ls3_final.groupby(['Property type', 'Room type'])['Price'].mean().reset_index(name='Average price')
print("\n--- Promedios agrupados calculados a partir de ls3_final (Top 10) ---")
print(ls3_final_promedios_agrupados.sort_values(by='Average price', ascending=False).head(10).to_string(index=False))
   
```

```python
# --- 1. IDENTIFICAR EL REGISTRO ESPECÍFICO A ELIMINAR ---
# Vamos a buscar el registro de Villa Soldati, Entire home/apt con el precio 2655.
villa_soldati_entire_apt_problem_row = ls3_final[
    (ls3_final['Neighbourhood'] == 'Villa Soldati') &
    (ls3_final['Room type'] == 'Entire home/apt') &
    (ls3_final['Price'] == 2655)
]

# Unir los índices de las filas problemáticas a eliminar
indices_to_remove_from_villa_soldati = pd.Index([])
if not villa_soldati_entire_apt_problem_row.empty:
    indices_to_remove_from_villa_soldati = indices_to_remove_from_villa_soldati.union(villa_soldati_entire_apt_problem_row.index)

print(f"Número de registros de Villa Soldati, Entire home/apt identificados para eliminar: {len(indices_to_remove_from_villa_soldati)}")
if not indices_to_remove_from_villa_soldati.empty:
    print("Índices a eliminar:", indices_to_remove_from_villa_soldati.tolist())
else:
    print("No se encontró el registro problemático de Villa Soldati, Entire home/apt.")


# --- 2. ELIMINAR ESE REGISTRO DE LS5_FINAL Y GUARDAR EL RESULTADO EN LS6 ---
ls4 = ls3_final.drop(indices_to_remove_from_villa_soldati)

# --- 3. VERIFICAR LOS RESULTADOS FINALMENTE ---
print(f"\n--- VERIFICACIÓN DESPUÉS DE LA ELIMINACIÓN EXPLÍCITA DE VILLA SOLDATI ENTIRE HOME ---")
print(f"Número de filas en ls3_final antes de esta eliminación: {len(ls3_final)}")
print(f"Número de filas en ls4: {len(ls4)}")
print(f"Filas eliminadas en este paso: {len(ls3_final) - len(ls4)}\n")
```

```python
def clasificar_actualizacion_corregida(valor):
    valor_limpio = valor.strip().lower()

    if valor_limpio == 'never':
        return 'Low'
    if valor_limpio == 'today':
        return 'High'

    total_dias = 0

    # Buscar días
    match_dias = re.search(r'(\d+)\s*day', valor_limpio)
    if match_dias:
        total_dias = int(match_dias.group(1))
    
    # Buscar semanas
    match_semanas = re.search(r'(\d+)\s*week', valor_limpio)
    if match_semanas:
        total_dias = int(match_semanas.group(1)) * 7

    # Buscar meses
    match_meses = re.search(r'(\d+)\s*month', valor_limpio)
    if match_meses:
        total_dias = int(match_meses.group(1)) * 30 # Aproximación
        
    # Buscar años (si aplica)
    match_years = re.search(r'(\d+)\s*year', valor_limpio)
    if match_years:
        total_dias = int(match_years.group(1)) * 365 # Aproximación

    # Ahora aplica una ÚNICA lógica a total_dias
    if total_dias <= 30: # Dentro del primer mes
        return 'High'
    elif total_dias <= 90: # Entre 1 y 3 meses
        return 'Medium'
    else: # Más de 3 meses
        return 'Low'

# Aplicar la función corregida
ls4['Segmenting the calendar update'] = ls4['Calendar updated'].apply(clasificar_actualizacion_corregida)
```

```python
def clasificar_revisiones(cantidad):
    if cantidad > 10:
        return 'High'
    elif cantidad >= 4:
        return 'Medium'
    else:
        return 'Low'

# Crear la nueva columna en ls5
ls4['Review count ranking'] = ls4['Number of reviews from the last 12 months'].apply(clasificar_revisiones)

# Verifica los primeros registros
print(ls4[['Number of reviews from the last 12 months', 'Review count ranking']].head())
```

```python
# Agrupar y contar las combinaciones entre 'Room type' y 'Neighbourhood'
edges = (
    ls4.groupby(['Room type', 'Neighbourhood'])
       .size()
       .reset_index(name='value')
)

# Eliminar filas con valores nulos, si hay
edges = edges.dropna()

# Convertir a string para evitar problemas
edges['source'] = edges['Room type'].astype(str)
edges['target'] = edges['Neighbourhood'].astype(str)

# Seleccionamos las columnas en orden para hv.Sankey
edges_formato = edges[['source', 'target', 'value']]

# Crear el Sankey diagram
sankey = hv.Sankey(edges_formato)

# Opciones de visualización
sankey.opts(
    label_position='left',
    edge_color='target',
    node_color='category',  # puedes cambiar a otro esquema si quieres
    cmap='tab20'
)

# Mostrar el gráfico
hv.output(sankey)
```

```python
# Agrupar por 'Room type' y 'Review count ranking' y contar
edges = (
    ls4.groupby(['Room type', 'Review count ranking'])
       .size()
       .reset_index(name='value')
)

# Eliminar filas con valores nulos, si hay
edges = edges.dropna()

# Convertir a string para evitar problemas
edges['source'] = edges['Room type'].astype(str)
edges['target'] = edges['Review count ranking'].astype(str)

# Seleccionar en orden
edges_formato = edges[['source', 'target', 'value']]

# Crear Sankey
sankey = hv.Sankey(edges_formato)

# Opciones de visualización, con título en inglés
sankey.opts(
    label_position='left',
    edge_color='target',
    node_color='category',  # opcional
    cmap='tab20',
    title='Relationship between Room Type and Review Count'
)

# Mostrar
hv.output(sankey)
```
